```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r'/Users/arseny/Documents/kaggle_datasets/heart.csv')
df = df[df['Oldpeak']>-2]
median_chol = df[df['Cholesterol'] > 0]['Cholesterol'].median()
df.loc[df['Cholesterol'] == 0, 'Cholesterol'] = median_chol
median_RestingBP= df[df['RestingBP']>0]['RestingBP'].median()
df.loc[df['RestingBP'] == 0, 'RestingBP'] = median_RestingBP
df.describe()
```

```python
cat_column = ['Sex', 'ChestPainType', 'RestingECG', 'ExerciseAngina', 'ST_Slope']
df = pd.get_dummies(df, columns=cat_column, drop_first=True)
df.info()
```

```python
df['age_oldpeak'] = df['Age'] * df['Oldpeak']
df['maxhr_age_ratio'] = df['MaxHR'] / df['Age']
```

```python
df.to_csv('/Users/arseny/Heart-Failure-Prediction/data/processed/heart_ready.csv', index=False)
```
